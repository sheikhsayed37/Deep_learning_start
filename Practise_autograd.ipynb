```python
import torch
x=torch.tensor([2.0], requires_grad=True)
y=x**2+3*x+1
y.backward()
x.grad
```

 ↓

dy/dx = 2x + 3

 ↓

x = 2 বসায়

 ↓

#2.What happens in this case?

 x = torch.tensor([2.0])

 y = x**2

 y.backward()


 Why does it fail?


```python

x = torch.tensor([2.0])

y = x**2

y.backward()
```

```python

x = torch.tensor([2.0],requires_grad= True)

y = x**2

y.backward()
x.grad
```

x = 2

│

│  requires_grad = False

↓

y = x²

│

│  No computation graph for gradients

↓

y.backward()

│

↓
❌ Error

Correct version

x = torch.tensor([2.0], requires_grad=True)

y = x**2


y.backward()


print(x.grad)

Output:

tensor([4.])

Because:

y=x**2

dx/
dy
	​=2x

At x=2:

dx/
dy=4


#Note

requires_grad=True tells PyTorch to track the tensor's operations and build a computation graph so that .backward() can calculate its gradients. Without it, gradient tracking is disabled, so y.backward() fails.
