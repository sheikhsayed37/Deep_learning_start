```python
import torch
x=torch.tensor(6.7)
y=torch.tensor(0.0)

w=torch.tensor(1.0)
b=torch.tensor(0.0)
```

```python
def binary_cross_entropy_loss(prediction,target):

  epsilon= 1e-8
  prediction=torch.clamp(prediction,epsilon,1-epsilon)

  loss = -(target * torch.log(prediction) +
             (1 - target) * torch.log(1 - prediction))

  return loss
```

```python
z=w*x+b
y_pred=torch.sigmoid(z)

loss = binary_cross_entropy_loss(y_pred, y)
```

```python
loss
```

```python
# Derivatives:
# 1. dL/d(y_pred): Loss with respect to the prediction (y_pred)
dloss_dy_pred = (y_pred - y)/(y_pred*(1-y_pred))

# 2. dy_pred/dz: Prediction (y_pred) with respect to z (sigmoid derivative)
dy_pred_dz = y_pred * (1 - y_pred)

# 3. dz/dw and dz/db: z with respect to w and b
dz_dw = x  # dz/dw = x
dz_db = 1  # dz/db = 1 (bias contributes directly to z)

dL_dw = dloss_dy_pred * dy_pred_dz * dz_dw
dL_db = dloss_dy_pred * dy_pred_dz * dz_db
```

```python
print(f"Manual Gradient of loss w.r.t weight (dw): {dL_dw}")
print(f"Manual Gradient of loss w.r.t bias (db): {dL_db}")
```

```python
# PYTORCH FOR GD
w = torch.tensor(1.0, requires_grad=True)
b = torch.tensor(0.0, requires_grad=True)

z = w * x + b
y_pred = torch.sigmoid(z)
loss = binary_cross_entropy_loss(y_pred, y)

loss.backward()

print(w.grad)
print(b.grad)
```

# Detach()
detach() creates a new tensor with the same values as the original, but this new tensor is outside the computation graph. That means:

The original tensor still tracks gradients.

The detached tensor does not track gradients.

You can use it for logging, plotting, or any operation where you don’t need backpropagation.


```python
import torch

x = torch.tensor(2.0, requires_grad=True)
z = x.detach()


y1 = x ** 2
y1.backward()
print("x.grad =", x.grad)
```

```python
z2 = z ** 2
z2.backward()
```

```python
y = x ** 2
y.backward()

print(x.grad)  # tensor(4.)
```

# torch.no_grad()

```python
# normal case
import torch

x = torch.tensor(2.0, requires_grad=True)
y = x ** 2

print(y.grad_fn)


y.backward()
print(x.grad)  # tensor(4.)
```

```python
import torch

x = torch.tensor(2.0, requires_grad=True)

with torch.no_grad():
    y = x ** 2

print(y.grad_fn)  # None
y.backward()      # error
```
